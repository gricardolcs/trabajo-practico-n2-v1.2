# Exploratory Data Analysis

This notebook is used for exploratory data analysis (EDA) on the smoking behavior dataset. The goal of EDA is to understand the dataset better and identify patterns related to smoking behavior.

## Importing Libraries
We will start by importing the necessary libraries.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set the aesthetic style of the plots
sns.set_style('whitegrid')
```

## Loading the Dataset
Next, we will load the dataset and take a look at its structure.

```python
data = pd.read_csv('../data/smoking_behavior.csv')  # Update with the correct path to your dataset
data.head()
```

## Data Overview
Let's check the basic statistics and information about the dataset.

```python
data.info()
data.describe()
```

## Visualizations
We will create some visualizations to understand the distribution of smoking behavior and its relationship with other features.

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='smoker', data=data)
plt.title('Distribution of Smokers vs Non-Smokers')
plt.xlabel('Smoker')
plt.ylabel('Count')
plt.show()
```

## Correlation Matrix
Let's visualize the correlation matrix to identify relationships between features.

```python
plt.figure(figsize=(12, 8))
correlation = data.corr()
sns.heatmap(correlation, annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

## Conclusion
In this notebook, we performed exploratory data analysis on the smoking behavior dataset. We visualized the distribution of smokers and non-smokers, as well as the correlation between different features. Further analysis and modeling can be conducted based on these insights.
